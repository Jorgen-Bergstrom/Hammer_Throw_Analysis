### Determine the Distance the Hammer is Thrown when No Drag

```python
import math
v = 19.0                     # velocity [m/s]
angle = math.radians(40.0)
g = 9.81                     # gravity [m/s^2]

v_vert = math.sin(angle) * v # vertical velocity

# v = a * t, so t = v / a
# v0 - g*t = 0  => t = v0 / g
t = 2 * v_vert / g
print(f"time = {t:.3f}")

d = math.cos(angle) * v * t
print(f"solution 1: distance={d:.3f} m")

d = math.sin(2*angle) * v**2 / g
print(f"solution 2: distance={d:.3f} m")
```

Distance thrown: $d = \displaystyle\frac{v^2}{g} \sin(2 \alpha)$
